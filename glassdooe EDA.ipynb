# **Project Summary -**

This project focused on analyzing the Glassdoor Jobs dataset and developing a machine learning model to predict company ratings based on various company and job-related features. The main objective of the project was to understand the factors influencing company ratings and build an accurate predictive model that could support business decision-making and employee satisfaction analysis.

The project started with data collection and understanding of the dataset, which contained information such as company names, job descriptions, ratings, industry, sector, revenue, company size, and founded year. Exploratory Data Analysis (EDA) was performed to understand the structure of the dataset, identify patterns, detect missing values, and visualize important relationships among features. Various visualizations such as correlation heatmaps, pair plots, histograms, and bar charts were used to analyze the dataset effectively.

During the preprocessing stage, several data cleaning techniques were applied. Missing values and invalid entries such as -1 values were handled appropriately. Outlier handling techniques were also performed to improve data quality. Feature engineering was applied to create new features such as Company_Age and Description_Length, which improved the predictive capability of the machine learning models. Categorical variables including Industry, Sector, Revenue, and Size were encoded using Label Encoding for model compatibility.

Since the dataset also contained textual information in the Job Description column, textual preprocessing techniques were implemented. These included lower casing, punctuation removal, stopword removal, whitespace cleaning, tokenization, textnormalization, URL removal, and removal of words containing digits. These preprocessing steps improved the quality and consistency of the textual data.

Three hypothetical statements were created based on insights obtained during EDA, and hypothesis testing was performed using appropriate statistical techniques such as Independent T-Test and Pearson Correlation Test. These tests helped determine whether significant relationships existed between different variables in the dataset.

Several machine learning models were implemented for predicting company ratings, including Linear Regression, Decision Tree Regressor, and Random Forest Regressor. The dataset was split into training and testing sets using an 80:20 ratio. Model performance was evaluated using evaluation metrics such as Mean Absolute Error (MAE), Mean Squared Error (MSE), Root Mean Squared Error (RMSE), and R² Score.

To improve model performance, hyperparameter optimization techniques such as GridSearchCV and RandomizedSearchCV with:

5-Fold Cross Validation

were applied. These techniques improved prediction accuracy, reduced overfitting, and enhanced model stability.

Among all the implemented models, the Random Forest Regressor achieved the best performance with lower error metrics and a higher R² score. Feature importance analysis showed that features such as Company_Age, Description_Length, Industry, Revenue, and Sector significantly influenced company rating predictions. This analysis also improved model explainability and business understanding.

Finally, the best-performing model was saved in .pkl format using the joblib library and successfully tested on unseen data for sanity checking. The project successfully demonstrated the complete machine learning workflow, including data preprocessing, feature engineering, model building, evaluation, optimization, and deployment preparation. The final model is capable of providing reliable company rating predictions and can support business intelligence, employee satisfaction analysis, and recruitment decision-making in real-world applications.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("glassdoor_jobs.csv")

# Dataset shape before cleaning
print("Before Cleaning:", df.shape)

# Remove duplicates
df = df.drop_duplicates()

# Fill missing values
df['Industry'] = df['Industry'].fillna('Unknown')
df['Sector'] = df['Sector'].fillna('Unknown')
df['Type of ownership'] = df['Type of ownership'].fillna('Unknown')

# Clean salary column
df['Salary Estimate'] = df['Salary Estimate'].str.replace('(Glassdoor est.)', '', regex=False)
df['Salary Estimate'] = df['Salary Estimate'].str.replace('$', '', regex=False)
df['Salary Estimate'] = df['Salary Estimate'].str.replace('K', '', regex=False)

# Create minimum salary column
df['Min Salary'] = df['Salary Estimate'].apply(
    lambda x: x.split('-')[0] if '-' in str(x) else np.nan
)

# Convert to numeric
df['Min Salary'] = pd.to_numeric(df['Min Salary'], errors='coerce')

# Remove null salary rows
df = df.dropna(subset=['Min Salary'])

# Final dataset shape
print("After Cleaning:", df.shape)

# Display first 5 rows
df.head()# Write your code to make your dataset analysis ready.# Write your code to make your dataset analysis ready.
````

#### Chart - 1

````python
# ==================================================
# HISTOGRAM CHART FOR SALARY PREDICTION PROJECT
# ==================================================

# Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
df = pd.read_csv("glassdoor_jobs.csv")

# ----------------------------------------
# CLEAN SALARY COLUMN
# ----------------------------------------

salary_col = 'Salary Estimate'

# Remove invalid salary rows
df = df[df[salary_col] != '-1']

# Convert to string
df[salary_col] = df[salary_col].astype(str)

# Remove unwanted text
df[salary_col] = df[salary_col].str.replace('$', '', regex=False)
df[salary_col] = df[salary_col].str.replace('K', '', regex=False)
df[salary_col] = df[salary_col].str.replace('(Glassdoor est.)', '', regex=False)
df[salary_col] = df[salary_col].str.replace('(Employer est.)', '', regex=False)

# Remove extra text
df[salary_col] = df[salary_col].str.split('(').str[0]

# Keep only salary ranges
df = df[df[salary_col].str.contains('-')]

# Minimum salary
df['min_salary'] = df[salary_col].apply(
    lambda x: int(
        x.split('-')[0]
        .replace('Employer Provided Salary:', '')
        .strip()
    )
)

# Maximum salary
df['max_salary'] = df[salary_col].apply(
    lambda x: int(
        x.split('-')[1]
        .split()[0]
        .strip()
    )
)

# Average salary
df['avg_salary'] = (
    df['min_salary'] + df['max_salary']
) / 2

# ----------------------------------------
# HISTOGRAM CHART
# ----------------------------------------

plt.figure(figsize=(10,6))

sns.histplot(
    df['avg_salary'],
    bins=30,
    kde=True
)

# Title
plt.title(
    "Distribution of Average Salaries",
    fontsize=16
)

# Labels
plt.xlabel(
    "Average Salary (K Dollars)",
    fontsize=12
)

plt.ylabel(
    "Number of Jobs",
    fontsize=12
)

# Grid
plt.grid(True)

# Show chart
plt.show()# Chart - 1 visualization code
````
